# 📈 Linear Regression Explanation and Hands-on Example

Welcome to the hands-on explanation notebook for **Linear Regression**! In this notebook, we will:
1. Generate a synthetic dataset representing a linear relationship with noise.
2. Implement **Linear Regression** using `scikit-learn`.
3. Implement **Linear Regression from scratch** using **Gradient Descent** to understand the underlying mathematics.
4. Visualize the training process and the final best-fit line.
5. Evaluate model performance using standard metrics: Mean Squared Error (MSE) and R-squared ($R^2$).

Let's start by importing the necessary libraries.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# Set seed for reproducibility
np.random.seed(42)
```

## 1. Data Generation

We will generate synthetic data based on the linear relation:
$$y = w \cdot x + b + \epsilon$$

Where:
*   True weight ($w$) = 2.5
*   True bias ($b$) = 1.0
*   $\epsilon$ is random Gaussian noise representing real-world measurement fluctuations.

```python
# Generate 100 random data points between 0 and 5
X = np.random.rand(100, 1) * 5
true_w = 2.5
true_b = 1.0
noise = np.random.randn(100, 1) * 0.8

y = true_w * X + true_b + noise

# Plot the generated data
plt.figure(figsize=(8, 5))
plt.scatter(X, y, color='blue', alpha=0.6, label='Data Points (y = 2.5x + 1 + noise)')
plt.xlabel('X (Independent Variable)')
plt.ylabel('y (Dependent Variable)')
plt.title('Synthetic Linear Data Generation')
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.show()
```

## 2. Linear Regression using Scikit-Learn

Let's fit a regression model using the industry-standard library `scikit-learn`. The `LinearRegression` class fits an ordinary least squares linear regression model.

```python
# Initialize and fit the model
model = LinearRegression()
model.fit(X, y)

# Retrieve the learned parameters
learned_w = model.coef_[0][0]
learned_b = model.intercept_[0]

print(f"True Weight: {true_w} | Learned Weight: {learned_w:.4f}")
print(f"True Bias: {true_b}   | Learned Bias: {learned_b:.4f}")

# Make predictions
y_pred_sklearn = model.predict(X)
```

## 3. Custom Linear Regression from Scratch (Gradient Descent)

To understand how weights are optimized, we will build a linear regression model from scratch. Recall the cost function (Mean Squared Error):
$$J(w, b) = \frac{1}{2m} \sum_{i=1}^{m} (h_{w,b}(x^{(i)}) - y^{(i)})^2$$

And the gradient updates:
$$w \leftarrow w - \alpha \frac{1}{m} \sum_{i=1}^{m} (h_{w,b}(x^{(i)}) - y^{(i)}) \cdot x^{(i)}$$
$$b \leftarrow b - \alpha \frac{1}{m} \sum_{i=1}^{m} (h_{w,b}(x^{(i)}) - y^{(i)})$$

Let's implement this iteratively in python.

```python
# Hyperparameters
learning_rate = 0.05
epochs = 200
m = len(X)

# Initialize weights and bias to 0
w_custom = 0.0
b_custom = 0.0

# Store history for visualization
cost_history = []
w_history = []
b_history = []

for epoch in range(epochs):
    # Hypothesis prediction
    y_pred_custom = w_custom * X + b_custom
    
    # Calculate Mean Squared Error (MSE)
    error = y_pred_custom - y
    cost = (1 / (2 * m)) * np.sum(error ** 2)
    cost_history.append(cost)
    w_history.append(w_custom)
    b_history.append(b_custom)
    
    # Compute gradients
    dw = (1 / m) * np.sum(error * X)
    db = (1 / m) * np.sum(error)
    
    # Update weights
    w_custom -= learning_rate * dw
    b_custom -= learning_rate * db
    
    # Log progress every 40 epochs
    if epoch % 40 == 0:
        print(f"Epoch {epoch:03d} | Cost: {cost:.4f} | w: {w_custom:.4f} | b: {b_custom:.4f}")

print(f"\nFinal learned weights from scratch:")
print(f"Learned Weight (w): {w_custom:.4f}")
print(f"Learned Bias (b): {b_custom:.4f}")
```

## 4. Visualizing Gradient Descent Convergence

Let's see how the regression line updates over epochs. Initially, the line starts at the bottom ($w=0, b=0$) and slowly fits to the data.

```python
plt.figure(figsize=(10, 6))
plt.scatter(X, y, color='blue', alpha=0.5, label='Data Points')

# Plot lines at different training steps
epochs_to_plot = [0, 5, 15, 30, 80, 199]
colors = ['red', 'orange', 'gold', 'lightgreen', 'cyan', 'green']

for epoch, color in zip(epochs_to_plot, colors):
    w_t = w_history[epoch]
    b_t = b_history[epoch]
    y_line = w_t * X + b_t
    plt.plot(X, y_line, color=color, linestyle='--', linewidth=1.5,
             label=f'Epoch {epoch} (w={w_t:.2f}, b={b_t:.2f})')

plt.xlabel('X')
plt.ylabel('y')
plt.title('Gradient Descent: Evolution of Regression Line')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

Let's also plot the cost curve (loss curve) to verify that optimization was smooth and converged.

```python
plt.figure(figsize=(8, 4))
plt.plot(cost_history, color='purple', linewidth=2)
plt.xlabel('Epoch')
plt.ylabel('Cost (MSE / 2)')
plt.title('Cost Reduction Over Training Epochs')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

## 5. Evaluation Metrics

To quantify how well our regression model fits the data, we use two metrics:
1.  **Mean Squared Error (MSE)**: The average squared difference between predictions and actual values. Lower is better.
2.  **Coefficient of Determination ($R^2$ Score)**: The proportion of variance in the dependent variable that is predictable from the independent variables. Ranges from 0 to 1 (1 indicates a perfect fit).

```python
# Predictions using sklearn
y_pred_sklearn = model.predict(X)
mse_sklearn = mean_squared_error(y, y_pred_sklearn)
r2_sklearn = r2_score(y, y_pred_sklearn)

# Predictions using our custom trained weights
y_pred_scratch = w_custom * X + b_custom
mse_scratch = mean_squared_error(y, y_pred_scratch)
r2_scratch = r2_score(y, y_pred_scratch)

print("--- Scikit-Learn Metrics ---")
print(f"MSE: {mse_sklearn:.4f}")
print(f"R2 Score: {r2_sklearn:.4f}")

print("\n--- Custom Scratch Metrics ---")
print(f"MSE: {mse_scratch:.4f}")
print(f"R2 Score: {r2_scratch:.4f}")
```

## 💡 Connection to Deep Learning & Computer Vision
*   In **object detection (like YOLO)**, predicting bounding box coordinates ($x, y, w, h$) is a regression problem.
*   Instead of a simple straight line, deep networks learn highly non-linear regression functions.
*   Instead of optimizing 2 variables ($w$ and $b$), deep networks optimize millions of parameters, using backpropagation to calculate gradients similar to the derivatives we computed above.
