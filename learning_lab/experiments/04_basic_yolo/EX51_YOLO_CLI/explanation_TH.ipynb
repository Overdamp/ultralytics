CLI สร้าง **กระบวนการ OS ใหม่** ต่อการเรียกแต่ละครั้ง (~0.5 วินาที; CUDA kernel recompile ครั้งแรก)

## โครงสร้าง CLI
```bash
yolo TASK MODE key=value ...
```
| Field | ค่าที่รับได้ |
|-------|------------|
| TASK | `detect`, `segment`, `classify`, `pose`, `obb` |
| MODE | `predict`, `train`, `val`, `export`, `track` |
